## 3. แคลคูลัส (Calculus)

### 3.1 อนุพันธ์ (Derivatives)

```python
# อนุพันธ์เชิงตัวเลข
def numerical_derivative(f, x, h=1e-5):
    """คำนวณอนุพันธ์เชิงตัวเลข"""
    return (f(x + h) - f(x - h)) / (2 * h)

# ตัวอย่าง: f(x) = x²
f = lambda x: x**2

x_values = [1, 2, 3, 4, 5]
print("f(x) = x²")
print("f'(x) = 2x")
print("\nx\tf'(x) exact\tf'(x) numerical")
print("-" * 45)
for x in x_values:
    exact = 2 * x
    numerical = numerical_derivative(f, x)
    print(f"{x}\t{exact}\t\t{numerical:.6f}")
```

### 3.2 การแสดงภาพการไล่ลงตามเกรเดียนต์ (Gradient Descent)

```python
# การไล่ลงตามเกรเดียนต์สำหรับ L(w) = (w-3)² (ตรงกับตัวอย่างในบทที่ 2)
def gradient_descent_1d(start, learning_rate, epochs):
    w = start
    history = [w]

    for _ in range(epochs):
        gradient = 2 * (w - 3)  # อนุพันธ์ของ (w-3)²
        w = w - learning_rate * gradient
        history.append(w)

    return history

# รันการไล่ลงตามเกรเดียนต์ด้วยพารามิเตอร์เดียวกับตัวอย่างในบท
w0 = 0
learning_rate = 0.1
epochs = 5
history = gradient_descent_1d(start=w0, learning_rate=learning_rate, epochs=epochs)

# พิมพ์ค่า w และ L(w) ทุกรอบ เพื่อเทียบกับตารางในบท
print("รอบ\tw\t\tL(w)")
print("-" * 40)
for t, w in enumerate(history):
    loss = (w - 3) ** 2
    print(f"{t}\t{w:.3f}\t\t{loss:.2f}")

# แสดงภาพ
w_range = np.linspace(-2, 8, 100)
y_range = (w_range - 3) ** 2

plt.figure(figsize=(10, 6))
plt.plot(w_range, y_range, 'b-', label='L(w) = (w-3)²')
plt.scatter(history, [(w - 3) ** 2 for w in history], c='red', s=50, zorder=5, label='เส้นทางการไล่ลงตามเกรเดียนต์')
plt.plot(history, [(w - 3) ** 2 for w in history], 'r--', alpha=0.5)
plt.xlabel('w')
plt.ylabel('L(w)')
plt.title('การไล่ลงตามเกรเดียนต์: หาค่าต่ำสุดของ L(w) = (w-3)²')
plt.legend()
plt.show()

print(f"\nค่าเริ่มต้น: w = {w0}")
print(f"หลัง {epochs} รอบ: w = {history[-1]:.6f}")
print(f"ค่าต่ำสุดที่แท้จริง: w = 3")
```

### 3.3 กฎลูกโซ่ (Chain Rule)

```python
# กฎลูกโซ่: d/dx[f(g(x))] = f'(g(x)) · g'(x)

# ตัวอย่าง: y = σ(x²) โดย σ คือซิกมอยด์
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

# y = sigmoid(x²)
# dy/dx = sigmoid'(x²) · 2x

def chain_rule_example(x):
    # Forward: y = sigmoid(x²)
    inner = x**2
    outer = sigmoid(inner)

    # Backward: dy/dx
    d_outer = sigmoid_derivative(inner)  # d(sigmoid)/d(inner)
    d_inner = 2 * x                       # d(inner)/d(x)
    dy_dx = d_outer * d_inner

    return outer, dy_dx

x_test = 2.0
y, dy = chain_rule_example(x_test)
print(f"x = {x_test}")
print(f"y = sigmoid(x²) = {y:.6f}")
print(f"dy/dx = {dy:.6f}")
```

## 4. ฟังก์ชันกระตุ้น (Activation Functions)

```python
# ฟังก์ชันกระตุ้น
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def softmax(x):
    exp_x = np.exp(x - np.max(x))  # เสถียรภาพเชิงตัวเลข
    return exp_x / np.sum(exp_x)

# อนุพันธ์
def sigmoid_deriv(x):
    s = sigmoid(x)
    return s * (1 - s)

def tanh_deriv(x):
    return 1 - np.tanh(x)**2

def relu_deriv(x):
    return np.where(x > 0, 1, 0)

# วาดกราฟ
x = np.linspace(-5, 5, 100)

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Sigmoid
axes[0, 0].plot(x, sigmoid(x), 'b-', label='Sigmoid')
axes[0, 0].plot(x, sigmoid_deriv(x), 'r--', label="อนุพันธ์")
axes[0, 0].set_title('Sigmoid')
axes[0, 0].legend()

# Tanh
axes[0, 1].plot(x, tanh(x), 'b-', label='Tanh')
axes[0, 1].plot(x, tanh_deriv(x), 'r--', label="อนุพันธ์")
axes[0, 1].set_title('Tanh')
axes[0, 1].legend()

# ReLU
axes[1, 0].plot(x, relu(x), 'b-', label='ReLU')
axes[1, 0].plot(x, relu_deriv(x), 'r--', label="อนุพันธ์")
axes[1, 0].set_title('ReLU')
axes[1, 0].legend()

# Leaky ReLU
axes[1, 1].plot(x, leaky_relu(x), 'b-', label='Leaky ReLU')
axes[1, 1].plot(x, np.where(x > 0, 1, 0.01), 'r--', label="อนุพันธ์")
axes[1, 1].set_title('Leaky ReLU')
axes[1, 1].legend()

plt.tight_layout()
plt.show()
```

## 5. สถิติและความน่าจะเป็น

### 5.1 สถิติเชิงพรรณนา

```python
# Descriptive Statistics
data = np.array([23, 25, 28, 30, 32, 35, 37, 40, 42, 45])

print("=== สถิติเชิงพรรณนา ===")
print(f"ข้อมูล: {data}")
print(f"\nค่าเฉลี่ย (Mean): {np.mean(data):.2f}")
print(f"มัธยฐาน (Median): {np.median(data):.2f}")
print(f"ส่วนเบี่ยงเบนมาตรฐาน (Std): {np.std(data):.2f}")
print(f"ความแปรปรวน (Variance): {np.var(data):.2f}")
print(f"ค่าต่ำสุด (Min): {np.min(data)}")
print(f"ค่าสูงสุด (Max): {np.max(data)}")
```

### 5.2 การแจกแจงปกติ (Normal Distribution)

```python
# การแจกแจงปกติ
def normal_pdf(x, mu, sigma):
    """ฟังก์ชันความหนาแน่นของความน่าจะเป็นสำหรับการแจกแจงปกติ"""
    return (1 / (sigma * np.sqrt(2 * np.pi))) * \
           np.exp(-0.5 * ((x - mu) / sigma)**2)

x = np.linspace(-5, 5, 100)

plt.figure(figsize=(10, 6))
for sigma in [0.5, 1, 2]:
    plt.plot(x, normal_pdf(x, 0, sigma), label=f'σ = {sigma}')

plt.xlabel('x')
plt.ylabel('ความหนาแน่นของความน่าจะเป็น')
plt.title('การแจกแจงปกติที่ค่า σ ต่างกัน')
plt.legend()
plt.show()
```

### 5.3 ซอฟต์แม็กซ์ (Softmax)

```python
# ซอฟต์แม็กซ์สำหรับการจำแนกหลายคลาส
def softmax(x):
    """ฟังก์ชันซอฟต์แม็กซ์แปลงลอจิตเป็นความน่าจะเป็น"""
    exp_x = np.exp(x - np.max(x))  # เสถียรภาพเชิงตัวเลข
    return exp_x / np.sum(exp_x)

# ตัวอย่าง
logits = np.array([2.0, 1.0, 0.1])
probs = softmax(logits)

print(f"ลอจิต: {logits}")
print(f"ความน่าจะเป็นจากซอฟต์แม็กซ์: {probs}")
print(f"ผลรวมของความน่าจะเป็น: {np.sum(probs):.6f}")

# แสดงภาพ
plt.figure(figsize=(8, 5))
plt.bar(['คลาส 1', 'คลาส 2', 'คลาส 3'], probs, color=['blue', 'green', 'red'])
plt.ylabel('ความน่าจะเป็น')
plt.title('ผลลัพธ์ของซอฟต์แม็กซ์')
plt.ylim(0, 1)
for i, p in enumerate(probs):
    plt.text(i, p + 0.02, f'{p:.3f}', ha='center')
plt.show()
```

## 6. แบบฝึกหัดการคำนวณ

### แบบฝึกหัดที่ 1: ผลคูณภายใน (Dot Product)
จงคำนวณผลคูณภายในของ:
- a = [3, 4, 5]
- b = [1, 2, 3]

```python
# เขียนโค้ดที่นี่
a = np.array([3, 4, 5])
b = np.array([1, 2, 3])

dot = np.dot(a, b)
print(f"a · b = {dot}")
print(f"คำนวณด้วยมือ: 3×1 + 4×2 + 5×3 = {3*1 + 4*2 + 5*3}")
```

### แบบฝึกหัดที่ 2: การคูณเมทริกซ์ (Matrix Multiplication)
จงคำนวณผลคูณเมทริกซ์:
- A = [[1, 2], [3, 4]]
- B = [[2, 0], [1, 3]]

```python
# เขียนโค้ดที่นี่
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 3]])

C = np.dot(A, B)
print(f"A × B =\n{C}")
```

### แบบฝึกหัดที่ 3: การไล่ลงตามเกรเดียนต์ (Gradient Descent)
จงใช้การไล่ลงตามเกรเดียนต์หาค่าต่ำสุดของ f(x) = x² + 4x + 3
โดยเริ่มจาก x = 5, อัตราการเรียนรู้ = 0.1, จำนวนรอบ = 3 พร้อมแสดงค่า x และ f(x) ทุกรอบ

```python
# เขียนโค้ดที่นี่
def f(x):
    return x**2 + 4*x + 3

def df(x):
    return 2*x + 4  # derivative

x = 5
lr = 0.1
epochs = 3
history = [x]

print(f"รอบ 0: x = {x:.4f}, f(x) = {f(x):.4f}")
for t in range(1, epochs + 1):
    x = x - lr * df(x)
    history.append(x)
    print(f"รอบ {t}: x = {x:.4f}, f(x) = {f(x):.4f}")

print(f"\nค่าต่ำสุดที่แท้จริง: x = -2 (จากการแก้สมการ 2x + 4 = 0)")
```

### แบบฝึกหัดที่ 4: ซอฟต์แม็กซ์ (Softmax)
จงคำนวณซอฟต์แม็กซ์ของลอจิต = [3.0, 1.0, 0.2]

```python
# เขียนโค้ดที่นี่
logits = np.array([3.0, 1.0, 0.2])
probs = softmax(logits)

print(f"ลอจิต: {logits}")
print(f"ความน่าจะเป็น: {probs}")
print(f"ผลรวม: {np.sum(probs):.6f}")
```

### แบบฝึกหัดที่ 5: กฎลูกโซ่ (Chain Rule)
จงคำนวณอนุพันธ์ของ y = ซิกมอยด์(2x + 1) เมื่อ x = 1

```python
# เขียนโค้ดที่นี่
x = 1

# Forward
inner = 2*x + 1
y = sigmoid(inner)

# Backward (Chain Rule)
d_sigmoid = sigmoid(inner) * (1 - sigmoid(inner))  # d(sigmoid)/d(inner)
d_inner = 2  # d(inner)/d(x)
dy_dx = d_sigmoid * d_inner

print(f"x = {x}")
print(f"y = sigmoid(2x + 1) = {y:.6f}")
print(f"dy/dx = {dy_dx:.6f}")
```
