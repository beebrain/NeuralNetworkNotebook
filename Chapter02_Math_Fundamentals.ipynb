## 3. แคลคูลัส (Calculus)

### 3.1 Derivatives (อนุพันธ์)

```python
# Numerical Derivative
def numerical_derivative(f, x, h=1e-5):
    """คำนวณอนุพันธ์เชิงตัวเลข"""
    return (f(x + h) - f(x - h)) / (2 * h)

# ตัวอย่าง: f(x) = x²
f = lambda x: x**2

x_values = [1, 2, 3, 4, 5]
print("f(x) = x²")
print("f'(x) = 2x")
print("\nx\tf'(x) exact\tf'(x) numerical")
print("-" * 45)
for x in x_values:
    exact = 2 * x
    numerical = numerical_derivative(f, x)
    print(f"{x}\t{exact}\t\t{numerical:.6f}")
```

### 3.2 Gradient Descent Visualization

```python
# Gradient Descent for f(x) = x²
def gradient_descent_1d(start, learning_rate, epochs):
    x = start
    history = [x]
    
    for _ in range(epochs):
        gradient = 2 * x  # derivative of x²
        x = x - learning_rate * gradient
        history.append(x)
        
    return history

# Run gradient descent
history = gradient_descent_1d(start=10, learning_rate=0.1, epochs=20)

# Visualize
x_range = np.linspace(-12, 12, 100)
y_range = x_range**2

plt.figure(figsize=(10, 6))
plt.plot(x_range, y_range, 'b-', label='f(x) = x²')
plt.scatter(history, [h**2 for h in history], c='red', s=50, zorder=5, label='Gradient Descent Path')
plt.plot(history, [h**2 for h in history], 'r--', alpha=0.5)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Gradient Descent: หาค่าต่ำสุดของ f(x) = x²')
plt.legend()
plt.show()

print(f"ค่าเริ่มต้น: x = 10")
print(f"หลัง 20 epochs: x = {history[-1]:.6f}")
print(f"ค่าต่ำสุดที่แท้จริง: x = 0")
```

### 3.3 Chain Rule (กฎลูกโซ่)

```python
# Chain Rule: d/dx[f(g(x))] = f'(g(x)) · g'(x)

# ตัวอย่าง: y = σ(x²) โดย σ คือ sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

# y = sigmoid(x²)
# dy/dx = sigmoid'(x²) · 2x

def chain_rule_example(x):
    # Forward: y = sigmoid(x²)
    inner = x**2
    outer = sigmoid(inner)
    
    # Backward: dy/dx
    d_outer = sigmoid_derivative(inner)  # d(sigmoid)/d(inner)
    d_inner = 2 * x                       # d(inner)/d(x)
    dy_dx = d_outer * d_inner
    
    return outer, dy_dx

x_test = 2.0
y, dy = chain_rule_example(x_test)
print(f"x = {x_test}")
print(f"y = sigmoid(x²) = {y:.6f}")
print(f"dy/dx = {dy:.6f}")
```

## 4. ฟังก์ชันกระตุ้น (Activation Functions)

```python
# Activation Functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def softmax(x):
    exp_x = np.exp(x - np.max(x))  # stability trick
    return exp_x / np.sum(exp_x)

# Derivatives
def sigmoid_deriv(x):
    s = sigmoid(x)
    return s * (1 - s)

def tanh_deriv(x):
    return 1 - np.tanh(x)**2

def relu_deriv(x):
    return np.where(x > 0, 1, 0)

# Plot
x = np.linspace(-5, 5, 100)

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Sigmoid
axes[0, 0].plot(x, sigmoid(x), 'b-', label='Sigmoid')
axes[0, 0].plot(x, sigmoid_deriv(x), 'r--', label="Derivative")
axes[0, 0].set_title('Sigmoid')
axes[0, 0].legend()

# Tanh
axes[0, 1].plot(x, tanh(x), 'b-', label='Tanh')
axes[0, 1].plot(x, tanh_deriv(x), 'r--', label="Derivative")
axes[0, 1].set_title('Tanh')
axes[0, 1].legend()

# ReLU
axes[1, 0].plot(x, relu(x), 'b-', label='ReLU')
axes[1, 0].plot(x, relu_deriv(x), 'r--', label="Derivative")
axes[1, 0].set_title('ReLU')
axes[1, 0].legend()

# Leaky ReLU
axes[1, 1].plot(x, leaky_relu(x), 'b-', label='Leaky ReLU')
axes[1, 1].plot(x, np.where(x > 0, 1, 0.01), 'r--', label="Derivative")
axes[1, 1].set_title('Leaky ReLU')
axes[1, 1].legend()

plt.tight_layout()
plt.show()
```

## 5. สถิติและความน่าจะเป็น

### 5.1 สถิติเชิงพรรณนา

```python
# Descriptive Statistics
data = np.array([23, 25, 28, 30, 32, 35, 37, 40, 42, 45])

print("=== สถิติเชิงพรรณนา ===")
print(f"ข้อมูล: {data}")
print(f"\nค่าเฉลี่ย (Mean): {np.mean(data):.2f}")
print(f"มัธยฐาน (Median): {np.median(data):.2f}")
print(f"ส่วนเบี่ยงเบนมาตรฐาน (Std): {np.std(data):.2f}")
print(f"ความแปรปรวน (Variance): {np.var(data):.2f}")
print(f"ค่าต่ำสุด (Min): {np.min(data)}")
print(f"ค่าสูงสุด (Max): {np.max(data)}")
```

### 5.2 Normal Distribution

```python
# Normal Distribution
def normal_pdf(x, mu, sigma):
    """Probability Density Function ของ Normal Distribution"""
    return (1 / (sigma * np.sqrt(2 * np.pi))) * \
           np.exp(-0.5 * ((x - mu) / sigma)**2)

x = np.linspace(-5, 5, 100)

plt.figure(figsize=(10, 6))
for sigma in [0.5, 1, 2]:
    plt.plot(x, normal_pdf(x, 0, sigma), label=f'σ = {sigma}')

plt.xlabel('x')
plt.ylabel('Probability Density')
plt.title('Normal Distribution with Different σ')
plt.legend()
plt.show()
```

### 5.3 Softmax Function

```python
# Softmax for Multi-class Classification
def softmax(x):
    """Softmax function แปลง logits เป็น probabilities"""
    exp_x = np.exp(x - np.max(x))  # numerical stability
    return exp_x / np.sum(exp_x)

# Example
logits = np.array([2.0, 1.0, 0.1])
probs = softmax(logits)

print(f"Logits: {logits}")
print(f"Softmax Probabilities: {probs}")
print(f"Sum of probabilities: {np.sum(probs):.6f}")

# Visualization
plt.figure(figsize=(8, 5))
plt.bar(['Class 1', 'Class 2', 'Class 3'], probs, color=['blue', 'green', 'red'])
plt.ylabel('Probability')
plt.title('Softmax Output')
plt.ylim(0, 1)
for i, p in enumerate(probs):
    plt.text(i, p + 0.02, f'{p:.3f}', ha='center')
plt.show()
```

### แบบฝึกหัดที่ 3: Gradient Descent
จงใช้ Gradient Descent หาค่าต่ำสุดของ f(x) = x² + 4x + 3
โดยเริ่มจาก x = 5, learning rate = 0.1, epochs = 50

```python
# เขียนโค้ดที่นี่
def f(x):
    return x**2 + 4*x + 3

def df(x):
    return 2*x + 4  # derivative

x = 5
lr = 0.1
history = [x]

for _ in range(50):
    x = x - lr * df(x)
    history.append(x)

print(f"ค่าต่ำสุดที่พบ: x = {x:.4f}")
print(f"ค่าต่ำสุดที่แท้จริง: x = -2 (จากการแก้สมการ 2x + 4 = 0)")
```

### แบบฝึกหัดที่ 4: Softmax

```python
# เขียนโค้ดที่นี่
logits = np.array([3.0, 1.0, 0.2])
probs = softmax(logits)

print(f"Logits: {logits}")
print(f"Probabilities: {probs}")
print(f"Sum: {np.sum(probs):.6f}")
```

### แบบฝึกหัดที่ 5: Chain Rule
จงคำนวณอนุพันธ์ของ y = sigmoid(2x + 1) เมื่อ x = 1

```python
# เขียนโค้ดที่นี่
x = 1

# Forward
inner = 2*x + 1
y = sigmoid(inner)

# Backward (Chain Rule)
d_sigmoid = sigmoid(inner) * (1 - sigmoid(inner))  # d(sigmoid)/d(inner)
d_inner = 2  # d(inner)/d(x)
dy_dx = d_sigmoid * d_inner

print(f"x = {x}")
print(f"y = sigmoid(2x + 1) = {y:.6f}")
print(f"dy/dx = {dy_dx:.6f}")
```
